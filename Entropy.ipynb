```python
# Entropy=−∑p(x)⋅log2​p(x)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
calc = {
    p: -p * np.log2(p) - (1 - p) * np.log2(1 - p) for p in np.linspace(0.01, 0.99, 99)
}
```

```python
prob = np.array(list(calc.keys()))
entr = np.array(list(calc.values()))
```

```python
# Set seaborn style
plt.style.use("dark_background")

# Generate probabilities from 0.01 to 0.99
prob = np.linspace(0.01, 0.99, 100)
entr = -prob * np.log2(prob) - (1 - prob) * np.log2(1 - prob)

# Create the plot
plt.figure(figsize=(12, 4))
plt.plot(prob, entr, color="skyblue", linewidth=1.5, label="Entropy Curve")
sns.kdeplot(
    prob, linestyle="dashed", label="Probability Distribution Plot", linewidth=1
)

# Highlight the max entropy point (at p=0.5)
max_entropy = -0.5 * np.log2(0.5) - 0.5 * np.log2(0.5)
plt.scatter(
    [0.5], [max_entropy], color="crimson", s=50, zorder=5, label="Max Entropy (p=0.5)"
)
plt.axvline(0.5, color="crimson", linestyle="dotted", linewidth=2)

# Annotation
plt.annotate(
    "Maximum Uncertainty\n(50-50 Class Balance)",
    xy=(0.5, max_entropy),
    xytext=(0.55, 0.65),
    arrowprops=dict(arrowstyle="->", color="white", lw=1.5),
    fontsize=12,
    bbox=dict(boxstyle="round,pad=0.3", fc="#212020A6", ec="white", lw=1),
)

# Labels and title
plt.title("Entropy vs Probability", fontsize=12, fontweight="bold")
plt.xlabel("Probability of Class 1 (p)", fontsize=12)
plt.ylabel("Entropy", fontsize=12)

# Grid and legend
plt.grid(True, linestyle="dotted", alpha=0.25)
plt.legend()
plt.tight_layout()
plt.savefig("Entropy_Prob.png")
plt.show()
```
