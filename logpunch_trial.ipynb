```python
from logpunch.customexception import TrackedException
from logpunch.customlogger import CustomLogger, LoggerConfig
```

```python
cfg = LoggerConfig(log_dir='logs')
logger = CustomLogger(config=cfg).get_logger()
```

```python
from pydantic import BaseModel, Field, validate_call, ValidationError
from typing import List, Annotated
```

```python
class Bank(BaseModel):
    user_name: str = Field(description='User name of the user')

    @validate_call
    def GetPassword(self,mobile_no:int = Field(description='Mobile Number of the User',strict=True)):
        return f"Hey {self.user_name}, Your Mobile Number is {mobile_no}"
```

```python
try:
    user1 = Bank(user_name='Gourav')
except ValidationError as e:
    err = TrackedException(message=e)
    logger.error(err)
    raise
```
